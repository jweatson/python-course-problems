## Problem 1 - Basic functions
### Background
A function is a portion of code that can be initially defined, and then re-used. For example, if you needed to perform the same calculation multiple times, rather than writing it out repeatedly, you can define the calculation as a function, and "call" the function instead.

In Python, a function has the following syntax:

```python3
def power(x,y):
  z = x**y
  return z
```

While this is a very simple function, you can see how a program is defined (`def`), given a name (`power`), given arguments (`x` and `y`), a calculation is performed, and then the value is returned.
