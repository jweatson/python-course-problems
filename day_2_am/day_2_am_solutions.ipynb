## Problem 1 - Basic Functions
Calculating a prime number in the form of a function in a naive way:

```python
def is_number_prime(number): # Defining a function, a reusable block of code that performs a certain task, def is used to define a function, the name of the function follows that, and inside the brackets are the function arguments, which are the inputs, in this case, we only need 1, the number we want to test
  if number < 2: # Prime numbers cannot be less than 2, this is a special, or "edge" case, which needs to be tested for
    return False # A return statement ends the function, and returns that value, that value can be a variable, or a literal value, in this case, it's False, since the number is not prime
  for i in range(2, number): # We now test every number from 2 up to the number itself, however, consider that there may be a better way of doing this
    if number % i == 0:
      return False
  return True

print(is_number_prime(1)) # 1 is not a prime number, this will return false
print(is_number_prime(2)) # 2 is the smallest prime number, this will return true
print(is_number_prime(3)) # 3 is a prime number, this will return true
print(is_number_prime(541)) # This is the 100th prime number, a good test case, will return true
```

However, a very basic improvement would be that in order to find a prime number you only have to count up to the _square root_ of the tested number! Improving code by minimizing the number of calculations it has to make for the same result is known as _optimization_.

```python
def is_number_prime(number):
  if number < 2:
    return False
  for i in range(2, int(number**0.5) + 1): # Change to the code that runs up to the square root of the tested number, recall the mathematical operation section, **2 squares a number, so **(1/2) or **0.5 is the square root, we add 1 to make sure that the test includes the square root of the number itself as, converting to an int rounds down
    if number % i == 0:
      return False
  return True

print(is_number_prime(1)) # 1 is not a prime number, this will return false
print(is_number_prime(2)) # 2 is the smallest prime number, this will return true
print(is_number_prime(3)) # 3 is a prime number, this will return true
print(is_number_prime(541)) # This is the 100th prime number, a good test case, will return true
```

Now that we have our function, it can be called in the contexts described in the task sheet.

```python
test_number = int(input("Enter a number to test if it is prime: ")) # Unlike previous usages of input, we are directly converting it to an integer, since that saves a step later on
if is_number_prime(test_number): # We can now use our function to test the number, since it returns a boolean value, we can use it directly in an if statement
  print(test_number,"is a prime number!")
else:
  print(f"{test_number} is not a prime number.") # This is another way of doing the same thing, using an f-string rather than multiple print arguments, this is a bit cleaner, but is functionally the same
```

```python
n = 2 # Since we are running a loop, we need a counter for the current number to test within the loop, we start at 2, since it is the smallest prime number
n_prime = 0 # Number of prime numbers found
while n_prime < 100: # Loop until we find 100 prime numbers, n_prime != 100 would also work, but this is more clear
  if is_number_prime(n):
    print(n)
    n_prime += 1 # Since a prime was found, increase the count of primes found
  n += 1
```

## Problem 2 - Modules 
Included in this folder is the module circle.py, we will import it, and then run the functions within

```python
import circles
radius = float(input("Enter the radius of the circle: "))
circumference = circles.calculate_circumference(radius) # We can now use the function in our module to calculate the circumference, similarly to calling a function, note that we have to include the module name
area = circles.calculate_area(radius) # Similarly, we can calculate the area
print("Your circle has a radius of {:.3f}, a circumference of {:.3f} and an area of {:.3f}".format(radius, circumference, area)) # This is another way of formatting a string, using the format method, the {:.3f} means to format the number as a float with 3 decimal places, this just makes the output look tidier
```

A modules functions can also be specifically imported, to keep things looking simpler, especially for a larger module.

```python
from circles import calculate_circumference, calculate_area # This is another way of importing from a module, we can specify which functions we want to import, this way we don't have to include the module name when calling the function
radius = float(input("Enter the radius of the circle: "))
circumference = calculate_circumference(radius)
area = calculate_area(radius)
print("Your circle has a radius of {:.3f}, a circumference of {:.3f} and an area of {:.3f}".format(radius, circumference, area))
```

## Problem 3 - Reading files
Writing a function to read numbers into a list (this doesn't have to be done as a function, but is a good idea when you are writing things in the future)

```python
def read_file(filename):
  file = open(filename, "r") # This variable is an open file, the "r" means read-only, it cannot write to the file
  contents = []
  for line in file: # We can loop through the file, each line is separately read in the form of a string
    number = int(line)
    contents.append(number)
  file.close() # It's good practice to close the file when we're done with it
  return contents

# Here is also a more compact way of doing the same thing, which will automatically close the file when we're done with it
def read_file(filename):
  contents = []
  with open(filename, "r") as file:
    for line in file:
      number = int(line)
      contents.append(number)
  return contents


numbers = read_file("random.txt")
sorted_numbers = sorted(numbers) # This is a built in function that sorts a list, we make a new variable to store the sorted list

# Now print the numbers line by line, we could also just print it as an array but this is neater
for number in sorted_numbers:
  print(number)
```
