## Problem 3 - Mathematical operations and types
Operators are used to perform operations on variables in Python, these can be mathematical or logical.

```python
x = 1
print(x)
x = x + 1 # The + operator can be used to add two numbers together
print(x) # Note that the value of x has changed after the addition operation.
x = x - 1 # Similarly, the - operator can be used to subtract one number from another.
print(x) # The value of x has changed again after the subtraction operation.
x = x * 2 # The * operator can be used to multiply two numbers together.
print(x) # The value of x has changed again after the multiplication operation.
x = x / 2 # The / operator can be used to divide one number by another.
print(x) # The value of x has changed again after the division operation.

# Multiple operations can be performed in a single line of code, and performed similarly to a mathematical equation, multiple variables can also be used in calculations, for instance:

x = 2
y = 3
z = x * y + 1 # This line of code performs multiple operations in a single line
print(z)

# More complex operations:
z = y % 2 # The % operator can be used to find the remainder of a division operation.
print(z) # The value of z is 1, since 3 divided by 2 leaves a remainder of 1.

z = x ** y # The ** operator can be used to perform exponentiation, which is raising a number to the power of another number.
print(z) # The value of z is 8, since 2 raised to the power of 3 is 8.
```

Which of the following are valid names for variables in Python?

```python
x = 1 # As we have already used x as a variable, we can assume it works
1 = 1 # This will not work, as you cannot assign a value to something that is itself a value
variable = 1 # This will work, as variable is a valid variable name
2variable = 1 # This will not work, as variable names cannot start with a number
variable2 = 1 # This will work, as variable names can contain numbers as long as they do not start with a number
variable_2 = 1 # This will work, as variable names can contain underscores
variable-2 = 1 # This will not work, as variable names cannot contain hyphens
variable$2 = 1 # This will not work, as variable names cannot contain special characters like $
global = 1 # This will work, but it is not recommended as global is used for other things
print = 1 # This will also work, but will break the print function, be careful about this!
```

What data type are the following values? We use the `type()` function to determine this, and can print it.

```python
print(type(23)) # 23 written like this is an integer
print(type(23.0)) # 23.0 written like this is becomes a float, which is a number with a decimal point, python will automatically convert integers into float if the mathematical operation requires it, for instance: 3/2 will give you 1.5, which is a float, even though 3 and 2 are integers
print(type("twenty three")) # "twenty three" is a string, which is a sequence of characters, strings are always enclosed in quotes, either single or double quotes will work, but they must match
print(type("23.5")) # "23.5" is also a string, even though it looks like a number, because it is enclosed in quotes, it is treated as a string, and not a number.
print(type(2.3e1)) # 2.3e1 is a float, but written in scientific notation, e is used in place of x10^, so 2.3e1 is the same as 2.3 x 10^1, which is 23.0
print(type(True)) # True is a boolean value, which can be either True or False, these are used to control how a program operates, and is used for conditions and loops, which we will learn about later.
print(type(true)) # This will give an error, because `true` is a variable that hasn't been defined yet, this is not the same as True
```

## Problem 4 - Inputs
This is the first program that will take multiple lines to execute, I am heavily commenting on what each line does, but when you write your own programs, it is better to keep your comments brief.

```python
celsius = input("Enter a number: ") # The input function allows a user to enter a value, the functions argument is a string that is displayed as a message to the user, this is optional, but a good idea to include as you want to be clear with the user what the program is doing
celsius = float(celsius) # Inputs are always treated as strings, so it needs to be converted, we use a float as it is easier to work with mathematically, integers are usually better for logical operations
print("Temperature in Celsius: ", celsius) # The print function can take multiple arguments, and will print them all out with a space in between, this is a good way to include a message with the value being printed out, there are other ways to do this, but this is the simplest one
fahrenheit = celsius * 1.8 + 32 # Convert celsius into fahrenheit, this could also be done in the form fahrenheit = (celsius * 9/5) + 32, but using 1.8 is simpler and more efficient
print("Temperature in Fahrenheit: ", fahrenheit) # Print the result of the conversion, again including a message to make it clear what is being printed out
```
