## Problem 1 - Logic and conditions
- `==` is the "equals" operator, if one variable is equal to the other, it returns `True`, if not, it returns `False`
- `!=` is the "not equals" operator, this does the opposite of  the "equals" operator
- `<` is the "less than" operator, for the case `x < y`, it returns `True` if `x` is less than `y`, and False if not
- `>` is the "greater than" operator, this does the opposite of the "less than" operator
- `<=` and `>=` are the "less than or equal to" and "greater than or equal to" operators, these behave similarly to the above operators, but are _also_ `True` when the values are equal

These operators will work on pretty much everything, but are mainly used in math, for instance, strings can be compared, such as determining if two string variables are exactly the same.

## Problem 2 - Loops
Write a program that counts up constantly using a while loop

```python
x = 1
while True: # Since this will always be the case, the program will remain within the while loop forever, this is called an infinite loop, and will run until the user stops it
  print(x)
  x += 1 # This is a shorthand way of writing x = x + 1, what I am doing here is called incrementing x, every time the loop runs back to the beginning it does the same thing again, with an increased value for x
  break # For the sake of not breaking this notebook, I have added a break statement, which exits out of a loop immediately, if you remove this statement, the program will run forever, and you will have to stop it manually. We won't need this for the next problem.
```

While infinite loops can be useful, introducing _conditions_ for the loop to end is far more useful and versatile when writing programs.

```python
x = 1
while x <= 10: # This is a condition, the loop will run as long as this condition is true, once it becomes false the loop will stop
  print(x)
  x += 1
# The program will now print 1 through 10, and then stop.
```

Using a `for` loop can be much more useful, since you do not need a counter, these can also be used to operate over a range or numbers, or a list, which we will get to later.

```python
for x in range(10): # We're now using a for loop instead, range is a function that generates a sequence of numbers, since comptuers count from 0, this will generate the numbers 0 through 9, every time the loop runs, x will be assigned the next number in the sequence, and the loop will run until it reaches the end of the sequence
  y = x + 1 # Since the loop is counting from 0, we need to add one
  print(y) # This will print the numbers 1 through 10, since we added 1 to x before printing it
```

The bonus question, creating a triangle of numbers:

```python
for x in range(10): # Now, we are using the range function with two arguments, the starting number, and the number it runs up to (this is not the final number, which would be 9 in this case)
  for y in range(x): # This is a nested loop, a loop within a loop, the inner loop will run every time the outer loop runs, and will keep running until the conditions of the outer loop are satisfied
    print(y+1, end=" ") # This will print the numbers from 1 to x, and the end=" " argument tells the print function to end with a space instead of a new line
  print() # This empty print statement will just make a new line at the end, so that the next time the loop runs, it will print on a new line

# Note that there is a single new line at the start of the output, this is because when x is originally 0, range(0) makes an empty sequence, so nothing is done
```

## Problem 3 - Multiple logical statements
This problem is commonly called a Fizz Buzz (https://en.wikipedia.org/wiki/Fizz_buzz#Programming)

```python
for n in range(1, 101): # remember, a computer counts from 0, since we're using two arguments in this loop, the first one is the starting number, and the final one is the number it runs up to, which is not included in the loop, so this will run from 1 to 100
  if n % 3 == 0 and n % 5 == 0: # The modulus operator gives us the remainder of a division, so if it is 0, then the number is evenly divisible by the other, for this statement, we are testing whether the number is divisible by both 3 and 5
    print("FizzBuzz")
  elif n % 3 == 0: # This is an elif statement, short for "else if", it is only tested if the first statement is false, so we know it is not divisible by both 3 and 5, so we test to see if it is divisible by 3
    print("Fizz")
  elif n % 5 == 0: # Similarly, but with 5
    print("Buzz")
  else: # This is an else statement, which covers all other cases not addressed by the previous statements, so this will always run if the number is not divisible by 3 or 5 or both
    print(n)
  # Now the loop runs again until we reach 100
```

## Problem 4 - Lists
A list is a type of variable that can contain other variables, 

```python
numbers = [] # We start with an empty list, a list can also be pre-filled with values, in the form [1,2,3] etc. but we will be adding values to this list using a loop
while True:
  x = input("Please enter a number:")
  if x == "": # If the user just presses enter, then x will be an empty string, and this will be the condition to stop the loop
    break
  numbers.append(float(x)) # This will add the number to the end of the list

# Now that we are eventually out of the list, we can do the rest of the problem
first_number = numbers[0] # This is how we access the first number in the list, since computers count from 0, the first number is at index 0, the second number is at index 1, and so on
print("The first number you entered was", first_number)
last_number = numbers[-1] # This is how we access the last number in the list
print("The last number you entered was", last_number)
min_number = min(numbers) # This is a built in function that gives us the minimum value in a list
print("The smallest number you entered was", min_number)
max_number = max(numbers) # This is a built in function that gives us the maximum valuein a list
print("The largest number you entered was", max_number)
len_numbers = len(numbers) # This is a built in function that gives us the length of a list, which is the number of items in the list
print("You entered", len_numbers, "numbers")
sum_numbers = sum(numbers) # This is a built in function that gives us the sum of all the numbers in a list
print("The sum of the numbers you entered was", sum_numbers)
mean = sum_numbers / len_numbers # Using the sum and the length, we can find the mean 
print("The mean of the numbers you entered was", mean)
```

### Bonus
In this case, we will break the input apart using the string `split` operator

```python
input_numbers = input("Please enter a list of numbers, separated by spaces:") # This will take the input as a single string
raw_numbers = input_numbers.split() # This will split the string into a list of strings, using spaces as the separator, keep in mind that this is still a list of strings

# Now, we can work through the list using a for loop to convert the strings to numbers, and store them in a new list
numbers = [] # We start with an empty list, and we will add the numbers to this list as we convert them
for num in raw_numbers: # This will loop through each item in the raw_numbers list, and assign it to the variable num, which we can then work with
  numbers.append(float(num)) # This will convert the string num to a float, and add it to the end of the numbers list

# Now we can do the same calculations as before, but with the new list of numbers
print(numbers) # Lists can also be printed, so this will contain all of the numbers entered

# The rest of this code is the same as before, so this is copied from the previous solution
first_number = numbers[0]
print("The first number you entered was", first_number)
last_number = numbers[-1]
print("The last number you entered was", last_number)
min_number = min(numbers)
print("The smallest number you entered was", min_number)
max_number = max(numbers)
print("The largest number you entered was", max_number)
len_numbers = len(numbers)
print("You entered", len_numbers, "numbers")
sum_numbers = sum(numbers)
print("The sum of the numbers you entered was", sum_numbers)
mean = sum_numbers / len_numbers
print("The mean of the numbers you entered was", mean)
```
