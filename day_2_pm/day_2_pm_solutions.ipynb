## Problem 1 - Writing files
As before, we write a function to read in files and sort the input file:

```python
def read_file(filename):
  contents = []
  with open(filename, "r") as file:
    for line in file:
      number = int(line)
      contents.append(number)
  return contents

numbers = read_file("random.txt")
sorted_numbers = sorted(numbers)
```

Now, we can make a new function to write the numbers to a file

```python
def write_file(filename, contents): # Note this function includes two arguments, the filename we want to save things to, and the contents, these have to be called in the correct order when you use the function
  with open(filename, "w") as file: # This is how we open a file for writing, the "w" argument means we are opening it for writing, if the file does not exist, it will be created, if it does exist, it will be overwritten (so be careful! It might be a good idea to look into how to check that a file exists and ask the user to conrfirm they want to overwrite it, for instance)
    for number in contents:
      file.write(str(number) + "\n") # The write method of a file object writes a string to the file, we convert the number back to a string, and add a new line character at the end so that there is a new line, this has to be done manually, unlike print
  # Since there is nothing to return, we can just leave the function without a return statement

write_file("sorted.txt", sorted_numbers) # Note that this is not being assigned to a variable, as there is nothing to return
```

## Problem 3 - `numpy`
First, we will make a program to calculate the square root of a lot of numbers using Python on its own

```python
import time

t1 = time.time()
numbers = []
for n in range(1,100000001):
  numbers.append(n)
roots = []
for number in numbers:
  roots.append(number**0.5)
t2 = time.time()
print("Time taken using python:", t2-t1, "seconds")
```

On my laptop, this took about 15 seconds. With Numpy:

```python
import numpy as np # Numpy is usually shortened to np as a common practice
t1 = time.time()
numbers = np.arange(1,100000001) # This creates an array of numbers from 1 to 100 million, no loop required
roots = np.sqrt(numbers) # This calculates all of the roots at once, no need to loop, and the operation is already optimized for you
t2 = time.time()
print("Time taken using numpy:", t2-t1, "seconds")
```

On my laptop, using numpy took 1.7 seconds, much, much faster! Numpy has lots of its own mathematical operations, you can also do math to an entire array, all at once, for instance, if you wanted to multiply all the roots by two and then add one you could do:

```python
t1 = time.time() 
roots_by_two = roots * 2 + 1
t2 = time.time()
print("Time taken to calculate roots*2+1 using numpy:", t2-t1, "seconds")
```

This is substantially faster than using a loop, again, and scales much better!

## Problem 4 - More Numpy
This is a very short problem, but gives you an idea of how useful numpy is! Rather than writing your own functions to read and write data, you can read it straight into an array.

```python
random_numbers = np.loadtxt("random.txt")
sorted_numbers = np.sort(random_numbers)
np.savetxt("sorted.txt", sorted_numbers, delimiter="\n") # A delimiter is something that is used to separate values, in this case we are using \n to separate the numbers with a new line character
```

As you can see we've written something that can do the entirety of the last task of the previous session and task 1 of this session into 3 lines of code, entirely using numpy, however, you may notice that the numbers are stored in scientific notation now, rather than strings, this is because numpy almost always operates using floats, and prefers to use scientific notation, you can change this using the `fmt` argument in the `savetxt` function.

## Bonus - Classes
Classes and objects are very high level topics, I don't expect you to need to use them! But I figured I'd include this since everything in python is some kind of class, and once you start noticing this, it makes a lot of aspects of Python easier.

```python
import math

class Circle: # We define the class similarly to a function, typically, we capitalize the name of a class to distinguish it from a function
  def __init__(self, radius): # The __init__ method is a special class function, or "method", that is called automatically when you create an object of this class
    self.radius = radius # The self keyword is a variable that is stored within the object, and is used to store attirbutes of the object, such as the radius of the circle
    # We can call other methods we define later in the class, and assign them as attributes too
    self.circumference = self.calculate_circumference()
    self.area = self.calculate_area()

  # Now, we define our other methods, they can be done in any order, but __self__ is usually put at the top, for the sake of readability, note that methods can use other functions outside of the class, such as math.pi, and that they operate identically to functions
  
  def calculate_circumference(self):
    return 2 * math.pi * self.radius
  def calculate_area(self):
    return math.pi * self.radius**2
  
radius = float(input("Enter the radius of the circle: "))
circle = Circle(radius)
print(f"The circumference of the circle is {circle.circumference}")
print(f"The area of the circle is {circle.area}")
```

You may notice that the way an attribute is called is similar to how functions are called by modules, this is because they are basically the same thing, just a more generalized form!
